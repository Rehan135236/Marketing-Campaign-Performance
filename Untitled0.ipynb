```python
# ============================================
# COMPLETE MARKETING DASHBOARD - AUTO DETECT COLUMNS
# ============================================

# 1. Setup and Imports
!pip install plotly -q

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

from google.colab import files
print("✅ Libraries imported successfully!")

# 2. Upload and Load Data
print("📤 Please upload your CSV file:")
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Load with encoding handling
try:
    df = pd.read_csv(filename, encoding='latin1')
    print(f"✅ Data loaded with latin1 encoding!")
except:
    df = pd.read_csv(filename, encoding='utf-8', engine='python')
    print(f"✅ Data loaded!")

print(f"📊 Shape: {df.shape[0]} rows × {df.shape[1]} columns")
print("\n📋 Column Names Found:")
print(df.columns.tolist())

# 3. Auto-detect and map column names
print("\n🔍 Detecting columns...")

# Find cost column
cost_col = None
for col in df.columns:
    if 'cost' in col.lower() and 'usd' in col.lower():
        cost_col = col
        break
    elif 'cost' in col.lower():
        cost_col = col

# Find revenue column
revenue_col = None
for col in df.columns:
    if 'revenue' in col.lower() and 'usd' in col.lower():
        revenue_col = col
        break
    elif 'revenue' in col.lower() or 'sales' in col.lower():
        revenue_col = col

# Find ROI column
roi_col = None
for col in df.columns:
    if col.lower() == 'roi' or 'roi' in col.lower():
        roi_col = col
        break

# Find channel column
channel_col = None
for col in df.columns:
    if 'channel' in col.lower():
        channel_col = col
        break

# Find impressions, clicks, leads, conversions
impressions_col = 'Impressions' if 'Impressions' in df.columns else None
clicks_col = 'Clicks' if 'Clicks' in df.columns else None
leads_col = 'Leads' if 'Leads' in df.columns else None
conversions_col = 'Conversions' if 'Conversions' in df.columns else None

print(f"  Cost column: {cost_col}")
print(f"  Revenue column: {revenue_col}")
print(f"  ROI column: {roi_col}")
print(f"  Channel column: {channel_col}")
print(f"  Impressions: {impressions_col}")
print(f"  Clicks: {clicks_col}")
print(f"  Leads: {leads_col}")
print(f"  Conversions: {conversions_col}")

# 4. Smart Data Cleaning
def clean_data(df):
    df_clean = df.copy()
    df_clean = df_clean.loc[:, ~df_clean.columns.str.contains('^Unnamed', case=False)]

    # Clean string columns - remove special characters
    for col in df_clean.select_dtypes(include=['object']).columns:
        df_clean[col] = df_clean[col].astype(str)
        df_clean[col] = df_clean[col].str.replace('\xa0', ' ', regex=False)
        df_clean[col] = df_clean[col].str.replace('Â', '', regex=False)
        df_clean[col] = df_clean[col].str.strip()

    # Clean monetary columns
    if cost_col:
        df_clean[cost_col] = df_clean[cost_col].astype(str)
        df_clean[cost_col] = df_clean[cost_col].str.replace('USD', '', regex=False)
        df_clean[cost_col] = df_clean[cost_col].str.replace('$', '', regex=False)
        df_clean[cost_col] = df_clean[cost_col].str.replace(',', '', regex=False)
        df_clean[cost_col] = df_clean[cost_col].str.strip()
        df_clean[cost_col] = pd.to_numeric(df_clean[cost_col], errors='coerce')
        print(f"✅ Cleaned {cost_col}")

    if revenue_col:
        df_clean[revenue_col] = df_clean[revenue_col].astype(str)
        df_clean[revenue_col] = df_clean[revenue_col].str.replace('USD', '', regex=False)
        df_clean[revenue_col] = df_clean[revenue_col].str.replace('$', '', regex=False)
        df_clean[revenue_col] = df_clean[revenue_col].str.replace(',', '', regex=False)
        df_clean[revenue_col] = df_clean[revenue_col].str.strip()
        df_clean[revenue_col] = pd.to_numeric(df_clean[revenue_col], errors='coerce')
        print(f"✅ Cleaned {revenue_col}")

    # Clean percentage columns
    pct_cols = ['Click Through Rate', 'Conversion Rate', 'Lead to Conversion Rate', 'CTR', 'CR']
    for col in pct_cols:
        if col in df_clean.columns:
            df_clean[col] = df_clean[col].astype(str)
            df_clean[col] = df_clean[col].str.replace('%', '', regex=False)
            df_clean[col] = df_clean[col].str.strip()
            df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce') / 100
            print(f"✅ Cleaned {col}")

    # Clean ROI
    if roi_col:
        df_clean[roi_col] = df_clean[roi_col].astype(str)
        df_clean[roi_col] = df_clean[roi_col].str.replace('%', '', regex=False)
        df_clean[roi_col] = df_clean[roi_col].str.replace('USD', '', regex=False)
        df_clean[roi_col] = df_clean[roi_col].str.strip()
        df_clean[roi_col] = pd.to_numeric(df_clean[roi_col], errors='coerce')
        if df_clean[roi_col].notna().any() and df_clean[roi_col].abs().max() > 10:
            df_clean[roi_col] = df_clean[roi_col] / 100
        print(f"✅ Cleaned {roi_col}")

    # Convert date columns if they exist
    if 'StartDate' in df_clean.columns:
        df_clean['StartDate'] = pd.to_datetime(df_clean['StartDate'], format='%d/%m/%Y', errors='coerce')
    if 'EndDate' in df_clean.columns:
        df_clean['EndDate'] = pd.to_datetime(df_clean['EndDate'], format='%d/%m/%Y', errors='coerce')

    return df_clean

df_clean = clean_data(df)
print("\n✅ Data cleaning completed!")

# 5. Check if needed columns exist
if not channel_col:
    print("\n⚠️ WARNING: No 'Channel' column found. Creating a default channel.")
    df_clean['Channel'] = 'Campaign'
    channel_col = 'Channel'

if not cost_col:
    print("⚠️ WARNING: No cost column found. Creating calculated cost from available data.")
    if 'Cost Per Acquisition' in df_clean.columns and conversions_col:
        df_clean['Cost_USD'] = df_clean['Cost Per Acquisition'] * df_clean[conversions_col]
        cost_col = 'Cost_USD'
    else:
        df_clean['Cost_USD'] = 1
        cost_col = 'Cost_USD'

if not revenue_col:
    print("⚠️ WARNING: No revenue column found. Creating calculated revenue.")
    df_clean['Revenue_USD'] = df_clean[cost_col] * 2  # Assume 2x ROI
    revenue_col = 'Revenue_USD'

if not roi_col:
    print("⚠️ WARNING: No ROI column found. Calculating from cost and revenue.")
    df_clean['ROI'] = df_clean[revenue_col] / df_clean[cost_col] - 1
    roi_col = 'ROI'

print(f"\n✅ Using columns:")
print(f"  Channel: {channel_col}")
print(f"  Cost: {cost_col}")
print(f"  Revenue: {revenue_col}")
print(f"  ROI: {roi_col}")

# 6. Prepare Aggregated Data
channel_data = df_clean.groupby(channel_col).agg({
    revenue_col: 'sum',
    cost_col: 'sum',
    roi_col: 'mean'
}).round(2)

# Rename columns for consistency
channel_data.columns = ['Revenue_USD', 'Cost_USD', 'ROI']

# Add impressions, clicks, etc. if available
if impressions_col:
    channel_data['Impressions'] = df_clean.groupby(channel_col)[impressions_col].sum()
if clicks_col:
    channel_data['Clicks'] = df_clean.groupby(channel_col)[clicks_col].sum()
if leads_col:
    channel_data['Leads'] = df_clean.groupby(channel_col)[leads_col].sum()
if conversions_col:
    channel_data['Conversions'] = df_clean.groupby(channel_col)[conversions_col].sum()

# Convert to millions for display
channel_data['Revenue_M'] = channel_data['Revenue_USD'] / 1_000_000
channel_data['Cost_M'] = channel_data['Cost_USD'] / 1_000_000

print("\n📊 Channel Data Summary:")
print(channel_data[['Revenue_M', 'Cost_M', 'ROI']].head())

# Monthly data if available
monthly_data = None
if 'Month' in df_clean.columns:
    month_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
    monthly_data = df_clean.groupby('Month').agg({
        revenue_col: 'sum',
        cost_col: 'sum'
    }).reindex(month_order)
    monthly_data.columns = ['Revenue_USD', 'Cost_USD']
    monthly_data['Revenue_M'] = monthly_data['Revenue_USD'] / 1_000_000
    monthly_data['Cost_M'] = monthly_data['Cost_USD'] / 1_000_000
    print("\n✅ Monthly data prepared")

# Funnel data
funnel_data = pd.DataFrame({
    'Stage': [],
    'Volume': []
})

if impressions_col:
    funnel_data = pd.concat([funnel_data, pd.DataFrame({'Stage': ['Impressions'], 'Volume': [df_clean[impressions_col].sum()]})])
if clicks_col:
    funnel_data = pd.concat([funnel_data, pd.DataFrame({'Stage': ['Clicks'], 'Volume': [df_clean[clicks_col].sum()]})])
if leads_col:
    funnel_data = pd.concat([funnel_data, pd.DataFrame({'Stage': ['Leads'], 'Volume': [df_clean[leads_col].sum()]})])
if conversions_col:
    funnel_data = pd.concat([funnel_data, pd.DataFrame({'Stage': ['Conversions'], 'Volume': [df_clean[conversions_col].sum()]})])

if len(funnel_data) > 0:
    funnel_data['Volume_M'] = funnel_data['Volume'] / 1_000_000
    print("\n✅ Funnel data prepared")

# 7. Create Dashboard Visuals
print("\n" + "="*60)
print("📊 GENERATING DASHBOARD VISUALS")
print("="*60)

# Create a large figure with subplots (2x3 grid)
fig = plt.figure(figsize=(20, 24))

# VISUAL 1: Revenue and Cost by Channel
ax1 = fig.add_subplot(3, 2, 1)
x = np.arange(len(channel_data.index))
width = 0.35
bars1 = ax1.bar(x - width/2, channel_data['Revenue_M'], width, label='Revenue', color='#2E86AB', alpha=0.8)
bars2 = ax1.bar(x + width/2, channel_data['Cost_M'], width, label='Cost', color='#A23B72', alpha=0.8)
ax1.set_xlabel('Channel', fontsize=11, fontweight='bold')
ax1.set_ylabel('Amount (Millions USD)', fontsize=11, fontweight='bold')
ax1.set_title('Revenue and Cost by Channel', fontsize=13, fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels(channel_data.index, rotation=45, ha='right')
ax1.legend(fontsize=10)
ax1.grid(axis='y', alpha=0.3)
for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        if height > 0:
            ax1.text(bar.get_x() + bar.get_width()/2., height, f'${height:.1f}M',
                    ha='center', va='bottom', fontsize=8)

# VISUAL 2: ROI by Channel
ax2 = fig.add_subplot(3, 2, 2)
channel_roi = channel_data['ROI'].sort_values()
colors = ['green' if x >= 1 else 'red' for x in channel_roi.values]
bars = ax2.barh(channel_roi.index, channel_roi.values, color=colors, alpha=0.7)
ax2.set_xlabel('Average ROI', fontsize=11, fontweight='bold')
ax2.set_title('ROI by Channel', fontsize=13, fontweight='bold')
ax2.axvline(x=1, color='red', linestyle='--', linewidth=2, label='Break-even (ROI=1)')
ax2.legend(loc='lower right', fontsize=10)
ax2.grid(axis='x', alpha=0.3)
for i, (bar, value) in enumerate(zip(bars, channel_roi.values)):
    ax2.text(value, bar.get_y() + bar.get_height()/2, f'{value:.2f}',
            ha='left' if value < max(channel_roi.values) else 'right',
            va='center', fontweight='bold', fontsize=10)

# VISUAL 3: ROI Contribution Pie Chart
ax3 = fig.add_subplot(3, 2, 3)
roi_contribution = channel_data['ROI']
roi_pct = (roi_contribution / roi_contribution.sum() * 100).round(1)
colors_pie = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4', '#FFEAA7']
wedges, texts, autotexts = ax3.pie(roi_pct, labels=roi_pct.index, autopct='%1.0f%%',
                                    colors=colors_pie, textprops={'fontsize': 10, 'fontweight': 'bold'},
                                    startangle=90)
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontweight('bold')
    autotext.set_fontsize(12)
ax3.set_title('ROI Contribution by Channel', fontsize=13, fontweight='bold', pad=20)

# VISUAL 4: Cost Distribution Pie Chart
ax4 = fig.add_subplot(3, 2, 4)
cost_pct = (channel_data['Cost_USD'] / channel_data['Cost_USD'].sum() * 100).round(1)
wedges, texts, autotexts = ax4.pie(cost_pct, labels=cost_pct.index, autopct='%1.0f%%',
                                    colors=colors_pie, textprops={'fontsize': 10, 'fontweight': 'bold'},
                                    startangle=90)
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontweight('bold')
    autotext.set_fontsize(12)
ax4.set_title('Cost Distribution by Channel', fontsize=13, fontweight='bold', pad=20)

# VISUAL 5: Monthly Trends
ax5 = fig.add_subplot(3, 2, 5)
if monthly_data is not None:
    ax5.plot(monthly_data.index, monthly_data['Revenue_M'], marker='o', linewidth=2,
             markersize=8, label='Revenue', color='#2E86AB')
    ax5.plot(monthly_data.index, monthly_data['Cost_M'], marker='s', linewidth=2,
             markersize=8, label='Cost', color='#A23B72')
    ax5.set_xlabel('Month', fontsize=11, fontweight='bold')
    ax5.set_ylabel('Amount (Millions USD)', fontsize=11, fontweight='bold')
    ax5.set_title('Monthly Revenue and Cost Trends', fontsize=13, fontweight='bold')
    ax5.legend(fontsize=10)
    ax5.grid(True, alpha=0.3)
    ax5.set_xticks(range(len(monthly_data.index)))
    ax5.set_xticklabels(monthly_data.index, rotation=45)
    for i, (month, row) in enumerate(monthly_data.iterrows()):
        if not pd.isna(row['Revenue_M']):
            ax5.text(i, row['Revenue_M'] + 0.02, f'{row["Revenue_M"]:.1f}M',
                    ha='center', va='bottom', fontsize=8)
        if not pd.isna(row['Cost_M']):
            ax5.text(i, row['Cost_M'] - 0.05, f'{row["Cost_M"]:.1f}M',
                    ha='center', va='top', fontsize=8)
else:
    ax5.text(0.5, 0.5, 'Month column not available\nAdd "Month" column for trend analysis',
            ha='center', va='center', transform=ax5.transAxes, fontsize=12)
    ax5.set_title('Monthly Revenue and Cost Trends', fontsize=13, fontweight='bold')

# VISUAL 6: Conversion Funnel
ax6 = fig.add_subplot(3, 2, 6)
if len(funnel_data) > 0:
    colors_funnel = ['#3498db', '#2ecc71', '#f39c12', '#e74c3c']
    bars = ax6.bar(funnel_data['Stage'], funnel_data['Volume_M'], color=colors_funnel[:len(funnel_data)], alpha=0.7)
    ax6.set_ylabel('Volume (Millions)', fontsize=11, fontweight='bold')
    ax6.set_title('Marketing Funnel - Volume by Stage', fontsize=13, fontweight='bold')
    ax6.set_yscale('log')
    ax6.grid(axis='y', alpha=0.3)
    for i, (bar, row) in enumerate(zip(bars, funnel_data.iterrows())):
        idx, row_data = row
        volume = row_data['Volume_M']
        ax6.text(bar.get_x() + bar.get_width()/2, volume, f'{volume:.1f}M',
                ha='center', va='bottom', fontweight='bold', fontsize=10)
        if i > 0:
            prev_volume = funnel_data.iloc[i-1]['Volume_M']
            if prev_volume > 0:
                rate = (volume / prev_volume * 100)
                ax6.text(i, volume/2, f'{rate:.1f}%', ha='center', va='center',
                        fontweight='bold', fontsize=11, color='white',
                        bbox=dict(boxstyle="round,pad=0.3", facecolor='black', alpha=0.7))
else:
    ax6.text(0.5, 0.5, 'Funnel data not available\nNeed Impressions, Clicks, Leads, Conversions',
            ha='center', va='center', transform=ax6.transAxes, fontsize=12)
    ax6.set_title('Marketing Funnel', fontsize=13, fontweight='bold')

plt.suptitle('📊 MARKETING CAMPAIGN PERFORMANCE DASHBOARD', fontsize=18, fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()

# 8. KPI Summary
print("\n" + "="*60)
print("📈 KEY PERFORMANCE INDICATORS (KPIs)")
print("="*60)

total_revenue = df_clean[revenue_col].sum()
total_cost = df_clean[cost_col].sum()
avg_roi = df_clean[roi_col].mean()
total_conversions = df_clean[conversions_col].sum() if conversions_col else 0
avg_cpa = total_cost / total_conversions if total_conversions > 0 else 0

# Determine ROI Status
if avg_roi < 0.8:
    roi_status = "Very Low"
elif avg_roi < 1:
    roi_status = "Low"
elif avg_roi < 1.5:
    roi_status = "Medium"
else:
    roi_status = "High"

kpi_data = pd.DataFrame({
    'Metric': ['Total Revenue', 'Total Cost', 'Average ROI', 'Total Conversions', 'Average CPA', 'ROI Status'],
    'Value': [f'${total_revenue/1_000_000:.2f}M', f'${total_cost/1_000_000:.2f}M',
              f'{avg_roi:.2f}', f'{total_conversions/1_000_000:.1f}M' if total_conversions > 0 else 'N/A',
              f'${avg_cpa:.2f}' if avg_cpa > 0 else 'N/A', roi_status]
})
print(kpi_data.to_string(index=False))

# 9. Insights
print("\n" + "="*60)
print("💡 KEY INSIGHTS")
print("="*60)
print(f"\n📊 Overall Performance:")
print(f"  • Total Revenue: ${total_revenue/1_000_000:.2f}M")
print(f"  • Total Cost: ${total_cost/1_000_000:.2f}M")
print(f"  • Profit: ${(total_revenue - total_cost)/1_000_000:.2f}M")
print(f"  • ROI: {avg_roi:.2f} - {'Profitable' if avg_roi >= 1 else 'Not Profitable'}")

if len(channel_data) > 0:
    best_channel = channel_data['ROI'].idxmax()
    worst_channel = channel_data['ROI'].idxmin()
    print(f"\n🎯 Channel Performance:")
    print(f"  • Best Channel: {best_channel} (ROI: {channel_data.loc[best_channel, 'ROI']:.2f})")
    print(f"  • Channel Needing Improvement: {worst_channel} (ROI: {channel_data.loc[worst_channel, 'ROI']:.2f})")

print("\n" + "="*60)
print("✅ DASHBOARD COMPLETED SUCCESSFULLY!")
print("="*60)
```
